## 3. Precision, Recall, and Confusion Matrix

In this section, you will train a logistic regression model to classify handwritten digits using a built-in sklearn dataset. You will then compute precision, recall, and a confusion matrix to understand where the model succeeds and where it fails.

Sklearn includes several ready-to-use datasets that require no external files. The `digits` dataset contains 1,797 grayscale images of handwritten digits (0-9), each stored as an 8x8 pixel array.

Source: adapted from the sklearn digits dataset example at https://scikit-learn.org/stable/auto_examples/classification/plot_digits_classification.html

```python
from sklearn.datasets import load_digits

digits = load_digits()
print('Images in dataset:', len(digits.images))
```

Each image is stored as an 8x8 array. The value at each cell represents pixel intensity: 0 corresponds to white and 16 (the maximum) corresponds to black. When building the classifier, we use the flattened 1x64 representation stored in `digits.data`.

The first image in the dataset:

```python
print('8x8 array of first image:')
print(digits.images[0])
```

```python
plt.figure(figsize=(2, 2))
plt.imshow(digits.images[0], cmap=plt.cm.gray_r, interpolation='nearest')
plt.title('Label: %i' % digits.target[0])
plt.show()
```

```python
# First 10 digits in the dataset
fig, axes = plt.subplots(1, 10, figsize=(20, 2))
for ax, (image, label) in zip(axes, list(zip(digits.images, digits.target))[:10]):
    ax.imshow(image, cmap=plt.cm.gray_r, interpolation='nearest')
    ax.set_title('Label: %i' % label)
    ax.axis('off')
plt.show()
```
